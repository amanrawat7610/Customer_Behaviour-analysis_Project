Data cleaning

```python
df['Item Purchased'].unique()
df['Category'].unique()
df.groupby('Category')['Item Purchased'].unique()

pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
df.groupby('Category')['Item Purchased'].unique()


correct_mapping = {

# Clothing

"T-shirt": "Clothing", "Shirt": "Clothing", "Shorts": "Clothing", "Hoodie": "Clothing", "Pants": "Clothing", "Socks": "Clothing", "Jeans": "Clothing", "Blouse": "Clothing", "Skirt": "Clothing", "Sweater": "Clothing", "Dress": "Clothing",



#Accessories

"Sunglasses": "Accessories", "Gloves": "Accessories", "Jewelry": "Accessories", "Hat": "Accessories", "Handbag": "Accessories", "Backpack": "Accessories", "Belt": "Accessories", "Scarf": "Accessories", "Bag": "Accessories",

#Electronics

"Laptop": "Electronics", "Phone": "Electronics", "Headphones": "Electronics", "watch": "Electronics",

#Footwear

"Shoes": "Footwear", "Sandals": "Footwear", "Sneakers": "Footwear", "Boots": "Footwear",

#Outerwear

"Coat": "Outerwear", "Jacket": "Outerwear"
}
correct_mapping
df.head()
df['Category']=df['Item Purchased'].map(correct_mapping)
df.groupby('Category')['Item Purchased'].unique()
```

Treating the null

size

```python
df['Size'].unique()
df.groupby('Category')['Size'].unique()
df.groupby('Item Purchased')['Size'].unique()
```

Treating null values and replacing all values in electronic category.

```python
df.loc[df['Category']=='Electronics','Size']="Not Applicable"
df.groupby('Category')['Item Purchased'].unique()
df.groupby('Category')['Size'].unique()
```

Treating null values and replacing all values in Accessories category.

```python
df.loc[df['Category']=='Accessories','Size']="free size"
df.groupby('Category')['Size'].unique()
```

Treating null values and replacing all values in Footwear category.

```python
df.loc[df['Category']=='Footwear','Size']="Not available"
df.groupby('Category')['Size'].unique()
```

Treating null values and replacing all values in Clothing category.

```python



df.loc[(df["Category"]=="Clothing") & (df["Size"].isnull()), "Size"] = df [df["Category"]=="Clothing"]["Size"].mode()[0]
df.groupby('Category')['Size'].unique()
df.isnull().sum()
df.loc[(df["Category"]=="Clothing") & (df["Size"].isnull()), "Size"] = df [df["Category"]=="Clothing"]["Size"].mode()[0]
df.isnull().sum()
df.groupby('Category')['Size'].unique()
# Ye check karega ki null values kahan kahan hain
df[df['Size'].isnull()]['Category'].value_counts()
```

Filling null values of Review Ratings with the mean of each Item Purchased

```python
df ['Review Rating']=df['Review Rating'].fillna(df.groupby('Item Purchased') ['Review Rating'] .transform ('mean'))
df.isnull().sum()
df.head(10)
df['Review Rating']=df['Review Rating'].round(2)
df.head(10)


df['Previous Purchases']=df['Previous Purchases'].fillna(0)
df.isnull().sum()

df ['Purchase Amount (USD)']=df['Purchase Amount (USD)'].fillna(df.groupby('Item Purchased') ['Purchase Amount (USD)'] .transform ('mean'))
df.isnull().sum()
df.head(20)




df['Size'] = df.groupby('Category')['Size'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'M'))
df.isnull().sum()


df.isnull().sum()
print(df.columns)
df = df.dropna(subset=['Category'])
```

Duplicates

```python
df.duplicated().sum()
len(df)
df[df.duplicated()]
df[df['Customer ID'] == 974]
df=df.drop_duplicates(subset=['Customer ID'],keep='first')
len(df)
df.count()

```

Renaming the headers

```python
df.columns=df.columns.str.replace(" ","_")

df.columns=df.columns.str.lower()
df.columns
```

rename columns purchase_amount_(usd) to purchase_amount

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
df.isnull().sum()

# Yeh check karein ki data ready hai
print("Columns:", df.columns)
print("Missing values total:", df.isnull().sum().sum())
# Agar sum 0 hai, toh data load karne ke liye ready hai.
```
